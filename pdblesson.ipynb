```python
%pip install -q biopython pandas matplotlib.
from Bio.PDB import PDBParser, PDBList
from Bio.SeqUtils import seq1

import pandas as pd
from pathlib import Path
project = Path("week02_mdm2_structure")

raw_folder = project / "data" / "raw"
processed_folder = project / "data" / "processed"
results_folder = project / "results"

raw_folder.mkdir(parents=True, exist_ok=True)
processed_folder.mkdir(parents=True, exist_ok=True)
results_folder.mkdir(parents=True, exist_ok=True)

print("Project folders created successfully.")
```

```python
from urllib.request import urlretrieve

pdb_id = "1YCR"

url = f"https://files.rcsb.org/download/{pdb_id}.pdb"

pdb_file = raw_folder / "1YCR.pdb"

urlretrieve(url, pdb_file)

print("Downloaded:", pdb_file)
```

```python
parser = PDBParser(QUIET=True)

structure = parser.get_structure(
    "MDM2_p53",
    str(pdb_file)
)

print("Structure ID:", structure.id)
```

```python
models = list(structure.get_models())

print("Number of models:", len(models))

model = models[0]

print("Selected model ID:", model.id)
```

```python
chains = list(model.get_chains())

print("Number of chains:", len(chains))

for chain in chains:
    print("Chain ID:", chain.id)
```

```python
for chain in chains:

    residues = list(chain.get_residues())
    atoms = list(chain.get_atoms())

    print(
        "Chain:", chain.id,
        "| Residues:", len(residues),
        "| Atoms:", len(atoms)
    )
```

```python
from Bio.PDB.Polypeptide import is_aa

for chain in chains:

    amino_acids = [
        residue
        for residue in chain.get_residues()
        if is_aa(residue, standard=True)
    ]

    sequence = "".join(
        seq1(residue.resname)
        for residue in amino_acids
    )

    print("Chain:", chain.id)
    print("Sequence:", sequence)
    print("Length:", len(sequence))
    print()
```

```python

summary_rows = []

for chain in chains:

    amino_acids = [
        residue
        for residue in chain.get_residues()
        if is_aa(residue, standard=True)
    ]

    sequence = "".join(
        seq1(residue.resname)
        for residue in amino_acids
    )

    summary_rows.append({
        "PDB_ID": pdb_id,
        "Chain": chain.id,
        "Residue_count": len(amino_acids),
        "Sequence": sequence
    })

summary_df = pd.DataFrame(summary_rows)

display(summary_df)
```
